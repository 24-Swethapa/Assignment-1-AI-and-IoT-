```python
 # ============================================================
# PART 1: DATASET LOADING AND EXPLORATION
# ============================================================

# pandas:
# Used for loading, cleaning and analyzing tabular IoT data.

# numpy:
# Used for numerical operations and creating the classification target.

import pandas as pd
import numpy as np

# train_test_split:
# Divides the dataset into training and testing data.

from sklearn.model_selection import train_test_split

# LinearRegression:
# Machine Learning algorithm used to predict continuous values.

# LogisticRegression:
# Machine Learning algorithm used to classify data into categories.

from sklearn.linear_model import LinearRegression, LogisticRegression

# Regression evaluation metrics:
# MAE, MSE and R² measure the performance of continuous-value prediction.

# Classification evaluation metrics:
# Accuracy, Precision, Recall and F1 Score measure classification performance.

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("All required libraries imported successfully!")
```

```python
# ------------------------------------------------------------
# LOAD THE IoT SENSOR DATASET
# ------------------------------------------------------------

# read_excel() loads the Excel dataset into a Pandas DataFrame.
# A DataFrame is a table-like structure used to store and analyze data.

df = pd.read_excel("/Folds5x2_pp.xlsx")

print("Dataset loaded successfully!")

```

```python
# head() displays the first 5 rows of the dataset.
# This helps us understand the structure and values of the data.

df.head()
```

```python
# shape returns:
# Number of rows = number of observations/data points
# Number of columns = number of attributes/features

rows, columns = df.shape

print("Number of observations:", rows)
print("Number of attributes:", columns)
```

```python
# info() displays:
# 1. Column names
# 2. Number of non-null values
# 3. Data types
# 4. Memory usage
#
# This helps us understand the structure of the dataset.

df.info()
```

```python
# describe() calculates important statistical values:
#
# Count  = number of observations
# Mean   = average value
# Std    = standard deviation
# Min    = minimum value
# 25%    = first quartile
# 50%    = median
# 75%    = third quartile
# Max    = maximum value
#
# Formula for Mean:
# Mean = Sum of all values / Number of values

df.describe()
```

```python
# isnull() identifies missing/empty values.
# sum() counts the missing values in each column.
#
# If the result is 0, that column contains no missing values.

missing_values = df.isnull().sum()

print("Missing values in each attribute:")
print(missing_values)

if missing_values.sum() == 0:
    print("\nResult: No missing values were found.")
else:
    print("\nResult: Missing values are present and need treatment.")
```

```python
# Numerical attributes contain numerical measurements.
# select_dtypes() identifies columns containing integer/float values.

numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

print("Numerical Attributes:")
print(list(numerical_columns))
```

```python
# Categorical attributes contain labels or categories such as:
# Male/Female, ON/OFF, Normal/Fault, etc.
#
# Here we check for object/category data types.

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

print("Categorical Attributes:")
print(list(categorical_columns))

if len(categorical_columns) == 0:
    print("\nResult: No categorical attributes are present.")
```

```python
# ============================================================
# PART 2: REGRESSION
# ============================================================

# REGRESSION:
# Regression is a supervised Machine Learning method used
# to predict a CONTINUOUS numerical value.
#
# In this dataset:
# AT = Ambient Temperature
# V  = Exhaust Vacuum
# AP = Ambient Pressure
# RH = Relative Humidity
#
# These four attributes are INPUT FEATURES.
#
# PE = Electrical Energy Output
# PE is the CONTINUOUS TARGET that we want to predict.

X = df[["AT", "V", "AP", "RH"]]
y = df["PE"]

print("Input Features:", X.columns.tolist())
print("Target Variable: PE")
```

```python
# ------------------------------------------------------------
# TRAIN-TEST SPLIT
# ------------------------------------------------------------

# The dataset is divided into:
# 80% → Training data
# 20% → Testing data
#
# Training data is used to teach the Machine Learning model.
# Testing data is used to evaluate the model on unseen data.
#
# test_size = 0.20 means 20% is reserved for testing.
#
# random_state = 42 ensures the same split is obtained
# every time the code is executed.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
# ------------------------------------------------------------
# LINEAR REGRESSION
# ------------------------------------------------------------

# Linear Regression finds a mathematical relationship between
# input features and a continuous target.
#
# General Linear Regression equation:
#
# y = β0 + β1X1 + β2X2 + ... + βnXn
#
# Where:
# y  = predicted output
# β0 = intercept
# β1...βn = model coefficients
# X1...Xn = input features
#
# Here:
#
# PE = β0 + β1(AT) + β2(V) + β3(AP) + β4(RH)

reg_model = LinearRegression()

# fit() trains the model using the training data.
reg_model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")

print("\nRegression Coefficients:")
print(reg_model.coef_)

print("\nIntercept:")
print(reg_model.intercept_)
```

```python
# ------------------------------------------------------------
# MAKE REGRESSION PREDICTIONS
# ------------------------------------------------------------

# predict() uses the trained Linear Regression model
# to predict PE values for the unseen test data.

y_pred = reg_model.predict(X_test)

print("Predictions generated successfully!")
print("\nFirst 10 predicted PE values:")
print(y_pred[:10])
```

```python
# ============================================================
# REGRESSION EVALUATION
# ============================================================

# 1. MAE - MEAN ABSOLUTE ERROR
#
# Formula:
#
#              1
# MAE =  ----------- Σ |Actual - Predicted|
#              n
#
# MAE gives the average absolute difference between
# actual and predicted values.
#
# Lower MAE = Better prediction.

mae = mean_absolute_error(y_test, y_pred)


# ------------------------------------------------------------
# 2. MSE - MEAN SQUARED ERROR
#
# Formula:
#
#              1
# MSE =  ----------- Σ (Actual - Predicted)²
#              n
#
# MSE squares the errors, so large errors have greater impact.
#
# Lower MSE = Better prediction.

mse = mean_squared_error(y_test, y_pred)


# ------------------------------------------------------------
# 3. R² - COEFFICIENT OF DETERMINATION
#
# Formula:
#
#              Σ(Actual - Predicted)²
# R² = 1 - ---------------------------
#              Σ(Actual - Mean)²
#
# R² indicates how well the model explains the variation
# in the target variable.
#
# R² closer to 1 = Better model fit.

r2 = r2_score(y_test, y_pred)


# ------------------------------------------------------------
# DISPLAY RESULTS
# ------------------------------------------------------------

print("REGRESSION EVALUATION RESULTS")
print("--------------------------------")
print("MAE :", mae)
print("MSE :", mse)
print("R²  :", r2)

print("\nMetric Interpretation:")
print("MAE → Lower is better")
print("MSE → Lower is better")
print("R²  → Closer to 1 is better")
```

```python
# Create a table containing all regression evaluation metrics.

regression_results = pd.DataFrame({
    "Metric": ["MAE", "MSE", "R²"],
    "Value": [mae, mse, r2]
})

regression_results
```

```python
# ============================================================
# PART 3: CLASSIFICATION
# ============================================================

# CLASSIFICATION:
# Classification is a supervised Machine Learning method
# used to predict a CATEGORY or CLASS.
#
# Our original PE value is continuous.
# Therefore, we convert PE into two categories:
#
# 0 → Low Energy
# 1 → High Energy
#
# We use the MEDIAN of PE as the threshold.
#
# Median = middle value when observations are arranged
# in ascending order.
#
# Classification rule:
#
# If PE >= median:
#       Energy_Class = 1 (High Energy)
#
# If PE < median:
#       Energy_Class = 0 (Low Energy)

threshold = df["PE"].median()

df["Energy_Class"] = np.where(
    df["PE"] >= threshold,
    1,
    0
)

print("PE Median / Classification Threshold:", threshold)

print("\nEnergy Class Distribution:")
print(df["Energy_Class"].value_counts())

print("\nClass Meaning:")
print("0 = Low Energy")
print("1 = High Energy")
```

```python
# ------------------------------------------------------------
# SELECT CLASSIFICATION INPUTS
# ------------------------------------------------------------

# We use the same four sensor/environmental features:
# AT, V, AP and RH.
#
# IMPORTANT:
# PE is NOT included as an input feature.
#
# Why?
# Because Energy_Class was created using PE.
# If PE were given to the model, it would already contain
# the information used to determine the answer.
#
# This problem is called TARGET LEAKAGE.
#
# Therefore:
#
# Inputs  → AT, V, AP, RH
# Target  → Energy_Class

X_class = df[["AT", "V", "AP", "RH"]]
y_class = df["Energy_Class"]

print("Classification Input Features:")
print(X_class.columns.tolist())

print("\nClassification Target:")
print("Energy_Class")
```

```python
# ------------------------------------------------------------
# CLASSIFICATION TRAIN-TEST SPLIT
# ------------------------------------------------------------

# 80% → Training data
# 20% → Testing data
#
# stratify=y_class maintains approximately the same proportion
# of Low Energy and High Energy observations in both datasets.

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class,
    y_class,
    test_size=0.20,
    random_state=42,
    stratify=y_class
)

print("Classification Training Samples:", len(X_train_c))
print("Classification Testing Samples:", len(X_test_c))
```

```python
# ------------------------------------------------------------
# LOGISTIC REGRESSION
# ------------------------------------------------------------

# Logistic Regression is used for CLASSIFICATION problems.
#
# It calculates the probability that an observation belongs
# to a particular class.
#
# Logistic/Sigmoid Function:
#
#              1
# P(Y=1) = ---------------
#          1 + e^(-z)
#
# where:
#
# z = β0 + β1X1 + β2X2 + ... + βnXn
#
# The probability is between 0 and 1.
#
# If probability >= 0.5:
#       Predict class 1 → High Energy
#
# If probability < 0.5:
#       Predict class 0 → Low Energy

clf_model = LogisticRegression(max_iter=1000)

# Train the Logistic Regression model.
clf_model.fit(X_train_c, y_train_c)

print("Logistic Regression model trained successfully!")
```

```python
# ------------------------------------------------------------
# MAKE CLASSIFICATION PREDICTIONS
# ------------------------------------------------------------

# predict() uses the trained Logistic Regression model
# to predict whether each observation belongs to:
#
# 0 → Low Energy
# 1 → High Energy

y_pred_c = clf_model.predict(X_test_c)

print("Classification predictions generated successfully!")

print("\nFirst 20 predictions:")
print(y_pred_c[:20])
```

```python
# ============================================================
# CLASSIFICATION EVALUATION
# ============================================================

# 1. ACCURACY
#
# Formula:
#
#                 TP + TN
# Accuracy = ---------------------
#             TP + TN + FP + FN
#
# TP = True Positive
# TN = True Negative
# FP = False Positive
# FN = False Negative
#
# Accuracy measures the percentage of all predictions
# that were correct.
#
# Higher Accuracy = Better.

accuracy = accuracy_score(y_test_c, y_pred_c)


# ------------------------------------------------------------
# 2. PRECISION
#
# Formula:
#
#               TP
# Precision = ---------
#              TP + FP
#
# Precision tells us:
# Of all observations predicted as High Energy,
# how many were actually High Energy?
#
# Higher Precision = Better.

precision = precision_score(y_test_c, y_pred_c)


# ------------------------------------------------------------
# 3. RECALL
#
# Formula:
#
#             TP
# Recall = ---------
#           TP + FN
#
# Recall tells us:
# Of all actual High Energy observations,
# how many were correctly identified?
#
# Higher Recall = Better.

recall = recall_score(y_test_c, y_pred_c)


# ------------------------------------------------------------
# 4. F1 SCORE
#
# Formula:
#
#                Precision × Recall
# F1 = 2 × ---------------------------
#                Precision + Recall
#
# F1 Score provides a balance between Precision and Recall.
#
# Higher F1 Score = Better.

f1 = f1_score(y_test_c, y_pred_c)


# ------------------------------------------------------------
# DISPLAY RESULTS
# ------------------------------------------------------------

print("CLASSIFICATION EVALUATION RESULTS")
print("-----------------------------------")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)

print("\nMetric Interpretation:")
print("Accuracy  → Higher is better")
print("Precision → Higher is better")
print("Recall    → Higher is better")
print("F1 Score  → Higher is better")
```

```python
# Create a table containing all classification metrics.

classification_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],
    "Value": [
        accuracy,
        precision,
        recall,
        f1
    ]
})

classification_results
```
